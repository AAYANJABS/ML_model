```python
import pandas as pd

rows = []
# (particle_size_um, flow_lpm, left_DE, right_DE)
table1 = [
    (10, 15, 25.81081, 45.27027),
    (10, 30, 32.97297, 59.45946),
    (10, 60, 32.02703, 67.36486),
    (5,  15, 16.48649, 27.63514),
    (5,  30, 20.60811, 35.06757),
    (5,  60, 26.89189, 47.09459),
    (1,  15, 12.90541, 24.18919),
    (1,  30, 15.27027, 27.97297),
    (1,  60, 16.68919, 31.75676),
]
for size, flow, left, right in table1:
    rows.append((size, flow, "left", left))
    rows.append((size, flow, "right", right))

df1 = pd.DataFrame(rows, columns=["size", "flow", "lung", "DE"])
print(df1)  # 18 rows
```

```python
X = pd.get_dummies(df1[["size", "flow", "lung"]], columns=["lung"])
y = df1["DE"]
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import LeaveOneOut, cross_val_predict
import numpy as np

rf = RandomForestRegressor(n_estimators=100, random_state=0)
pred = cross_val_predict(rf, X, y, cv=LeaveOneOut())

rmse = np.sqrt(np.mean((pred - y) ** 2))
mae = np.mean(np.abs(pred - y))
mbe = np.mean(pred - y)
r = np.corrcoef(pred, y)[0, 1]
print(f"RMSE={rmse:.3f} MAE={mae:.3f} MBE={mbe:.3f} R={r:.3f}")
```

```python
new = pd.DataFrame({"size": [2.5, 2.5], "flow": [25, 25], "lung_left": [1, 0], "lung_right": [0, 1]})
print(rf.fit(X, y).predict(new))
```
